1. What is a Decision Tree, and how does it work in the context of
classification?
  - Decision tree is an ML algorithm used to make descision according to features available. It keeps going until every node is a pure node.

2. Explain the concepts of Gini Impurity and Entropy as impurity measures. How do they impact the splits in a Decision Tree?
  - Gini and entrophy are used to measure the impurity in node. We start splitting from the node which has the lowest gini or entrophy impurity. 0 means pure node.

3. What is the difference between Pre-Pruning and Post-Pruning in Decision Trees? Give one practical advantage of using each.
  - Pre-Pruning means doing hyperparameter tuning, then creating the best model.
  - Post-Pruning means changing default parameters after visualizing the decision tree.
  - Pre-Pruning helps find the best from start, while in post-pruning we change parameters manually and see accuracy.

4. What is Information Gain in Decision Trees, and why is it important for choosing the best split?
  - Information Gain is difference between gini or entrophy of root node and combined gini or entrophy of leaf nodes.
  - The feature that gets highest IG is used for split.

5. What are some common real-world applications of Decision Trees, and what are their main advantages and limitations?
  - Detecting cancer, weather forecasting.
  - Advantages
    - Easy to understand & interpret (if-else rules)
    - Works with numeric & categorical data
    - Little data preprocessing (no scaling needed)
    - Handles non-linear relationships
    - Fast training & inference
  - Limitations
    - Overfitting (especially deep trees)
    - Unstable (small data changes → different tree)
    - Lower accuracy than ensembles (Random Forest, XGBoost)
    - Biased toward features with more splits
    - Poor extrapolation beyond seen data

```python
# 6. Load the Iris Dataset
# ● Train a Decision Tree Classifier using the Gini criterion
# ● Print the model’s accuracy and feature importances

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load the Iris dataset

iris_df = load_iris()
X = iris_df.data
y = iris_df.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
model = DecisionTreeClassifier(criterion="gini")
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
model.feature_importances_
```

```python
# 7. Load the Iris Dataset
# ● Train a Decision Tree Classifier with max_depth=3 and compare its accuracy to
# a fully-grown tree.

model = DecisionTreeClassifier(criterion="gini", max_depth=3)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
# 8. Load the Boston Housing Dataset
# ● Train a Decision Tree Regressor
# ● Print the Mean Squared Error (MSE) and feature importances

from sklearn.datasets import fetch_california_housing
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

boston_df = fetch_california_housing()
X = boston_df.data[:1000]
y = boston_df.target[:1000]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
model = DecisionTreeRegressor()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
mean_squared_error(y_test, y_pred)
```

```python
# 9. Load the Iris Dataset
# ● Tune the Decision Tree’s max_depth and min_samples_split using
# GridSearchCV
# ● Print the best parameters and the resulting model accuracy

from sklearn.model_selection import GridSearchCV

X = iris_df.data
y = iris_df.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

params = {"max_depth": [1,2,3,4,5,10,None],
          "min_samples_split": [2,3,4,5,10,15,20]}

grid_cv = GridSearchCV(DecisionTreeClassifier(), params, scoring="accuracy", cv=5)
grid_cv.fit(X_train, y_train)
grid_cv.best_params_
```

```python
y_pred = grid_cv.best_estimator_.predict(X_test)
accuracy_score(y_test, y_pred)
```
