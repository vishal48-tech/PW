15. What is causation? Explain difference between correlation and causation with an example.
  - Causation means one event directly causes another event to happen.
  - Correlation: Two things happen together (relationship), but one does NOT necessarily cause the other.
  - Causation: One thing directly causes the other.
  - E.g. When ice cream sales increase, drowning incidents also increase. Both increased because of summer. Drowning doesn't increased because of ice cream sales.

16. What is an Optimizer? What are different types of optimizers? Explain each with an example.
  - An optimizer is an algorithm used in Machine Learning to update the model’s parameters (weights and biases) so that the loss decreases.
  - Types of Optimizers:
    - Gradient Descent: It computes the gradient using all training samples and updates weights. E.g. Learning to classify cats vs. dogs using the entire dataset for each update.
    - Stochastic Gradient Descent: Updates weights one sample at a time. E.g. Takes 1 image → calculates gradient → updates weights.
    - Mini-Batch Gradient Descent: Updates weights using small batches (like 32, 64 images at a time). E.g. Batch size = 32, Take 32 images → compute average gradient → update weights.
    - Momentum Optimizer: Adds a “momentum” term → smooths out noisy updates. E.g. If previous updates were moving left, momentum continues moving left even if gradient fluctuates.
