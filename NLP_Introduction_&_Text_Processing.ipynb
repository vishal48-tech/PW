1.  What is Computational Linguistics and how does it relate to NLP?
- Computational linguistics uses computers to model and study how human language works as a cognitive and social system.
- NLP transforms raw data into machine learning dataset.

2. Briefly describe the historical evolution of Natural Language Processing.
- NLP began in the 1950s with hand-written rules for limited tasks like translation. The 1990s shifted to statistical models learning from data. The 2010s brought neural networks, and the 2020s revolutionized the field with large transformers (e.g., BERT, GPT), enabling fluid, context-aware language understanding.

3.  List and explain three major use cases of NLP in today's tech industry.
- Sentiment analysis: These helps us understand either the text is positive, negative or neutral.
- Intelligent search and information retrieval: Goes beyond keyword matching to understand user intent. Search engines can answer questions like `best quiet places to work near me` by interpreting meaning, context, and location.
- Language translation: Translating text from one language to another.

4. What is text normalization and why is it essential in text processing tasks?
- Text normalization is the process of transforming text into a consistent canonical form. It includes, cleaning, lowercasing, expanding contractions (e.g. 'don't' -> 'do not'), standardizing numbers and removing punctuation.
- It is essential because raw text is noisy and inconsistent. Normalization reduces irrelevant variation, allowing models to treat semantically equivalent tokens (e.g. 'Go', 'go', 'GO') as identical, improving accuracy and efficiency in downstream tasks like search or machine translation.

5. Compare and contrast stemming and lemmatization with suitable
examples.
- Stemming crudely chops off prefixes/suffixes, often yielding non words. Example. Studies -> studi.
- Lemmatization uses vocabulary and morphological analysis to return a valid base word. Example studies -> study.
- Stemming is faster but less accurate. Lemmatization is slower but reserves meaning.

```python
""" 7. Given the sample paragraph below, perform string tokenization and
frequency distribution using Python and NLTK:

“Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical.” """

import nltk
import string
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.probability import FreqDist

text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical."""

def clean_text(raw_text):
  text_lower = raw_text.lower()
  clean_text = text_lower.translate(str.maketrans("", "", string.punctuation))
  return clean_text

text = clean_text(text)
sentences = sent_tokenize(text)
words = [word_tokenize(sentence) for sentence in sentences]
fdist = FreqDist(words[0])
print(fdist.most_common())
```

```python
""" Create a custom annotator using spaCy or NLTK that identifies and labels
proper nouns in a given text.

“Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical.” """

import spacy
from spacy.tokens import Span

class ProperNounAnnotator:
    def __init__(self):
        # Load English model
        self.nlp = spacy.load("en_core_web_sm")

    def annotate(self, text):
        """Identify and label proper nouns in text."""
        doc = self.nlp(text)

        annotated_text = text
        proper_nouns = []

        # Collect proper nouns (PROPN tag)
        for token in doc:
            if token.pos_ == "PROPN":
                proper_nouns.append({
                    'word': token.text,
                    'position': token.i,
                    'start_char': token.idx,
                    'end_char': token.idx + len(token.text)
                })

        return proper_nouns, doc

# Your text
text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical."""

# Use the annotator
annotator = ProperNounAnnotator()
proper_nouns, doc = annotator.annotate(text)

print("PROPER NOUNS FOUND:")
for pn in proper_nouns:
    print(f"  • '{pn['word']}' at position {pn['position']}")

# Visualize with POS tags
print("\nTOKEN-LEVEL ANALYSIS:")
for token in doc:
    if token.pos_ == "PROPN":
        print(f"  {token.text} → [{token.pos_}] (Proper Noun)")
    elif token.pos_ == "NOUN":
        print(f"  {token.text} → [{token.pos_}] (Common Noun)")
```

```python
""" 9.  Using Genism, demonstrate how to train a simple Word2Vec model on the
following dataset consisting of example sentences:

dataset = [
"Natural language processing enables computers to understand human language",
"Word embeddings are a type of word representation that allows words with similar
meaning to have similar representation",
"Word2Vec is a popular word embedding technique used in many NLP applications",
"Text preprocessing is a critical step before training word embeddings",
"Tokenization and normalization help clean raw text for modeling"
]

Write code that tokenizes the dataset, preprocesses it, and trains a Word2Vec model using
Gensim."""

import gensim
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess, tokenize

dataset = [
        "Natural language processing enables computers to understand human language",
        "Word embeddings are a type of word representation that allows words with similar meaning to have similar representation",
        "Word2Vec is a popular word embedding technique used in many NLP applications",
        "Text preprocessing is a critical step before training word embeddings",
        "Tokenization and normalization help clean raw text for modeling"
    ]

def simple_word2vec_training(dataset):
    """
    Simpler approach with Gensim's built-in utilities
    """

    # Tokenize and preprocess
    processed_data = []
    for sentence in dataset:
        # Gensim's tokenizer (lowercases, removes punctuation)
        tokens = [token for token in tokenize(sentence.lower(), deacc=True)]
        processed_data.append(tokens)

    # Train Word2Vec
    model = Word2Vec(
        sentences=processed_data,
        vector_size=50,
        window=5,
        min_count=1,
        workers=4,
        epochs=10
    )

    return model, processed_data
```

```python
print("\n" + "=" * 60)
print("📝 : Simple Gensim Tokenizer")
print("-" * 40)

simple_model, processed_data = simple_word2vec_training(dataset)
print(f"✅ Model trained with {len(simple_model.wv)} words in vocabulary")

# Demonstrate model usage
print("\n" + "=" * 60)
print("🔍 EXPLORING THE WORD2VEC MODEL")
print("=" * 60)

# Get vocabulary
print(f"\n📖 Vocabulary words: {simple_model.wv.index_to_key}")

# Find similar words (if enough data)
if 'language' in simple_model.wv:
    print(f"\n🔗 Words similar to 'language':")
    similar_words = simple_model.wv.most_similar('language', topn=3)
    for word, score in similar_words:
        print(f"  • {word}: {score:.4f}")

# Get word vector
if 'word' in simple_model.wv:
    word_vector = simple_model.wv['word']
    print(f"\n📊 Vector for 'word':")
    print(f"  Shape: {word_vector.shape}")
    print(f"  First 5 dimensions: {word_vector[:5]}")

# Find word similarity
if 'natural' in simple_model.wv and 'language' in simple_model.wv:
    similarity = simple_model.wv.similarity('natural', 'language')
    print(f"\n🎯 Similarity between 'natural' and 'language': {similarity:.4f}")
```
