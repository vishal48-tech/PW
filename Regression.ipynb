1. What is Simple Linear Regression?
  - Simple Linear Regression contains only one independent variable.

2. What are the key assumptions of Simple Linear Regression?
  - There should be only one independent variable.
  - Relationship between x and y should be linear.
  - Constant variance across errors of x.
  - No outliers.

3. What does the coefficient m represent in the equation Y=mX+c?
  - coefficient m represents slope of the line in the equation Y=mX+c.

4. What does the intercept c represent in the equation Y=mX+c?
  - intercept c represents how much far is line from origin(0,0) in the equation Y=mX+c.

5. How do we calculate the slope m in Simple Linear Regression?
  - (y2-y1)/(x2-x1)

6. What is the purpose of the least squares method in Simple Linear Regression?
  - This method finds the line with least error.

7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?
  - R² represents the proportion of variance in the dependent variable (Y) that is explained by the independent variable (X) through the regression model.

8. What is Multiple Linear Regression?
  - Multiple Linear Regression contains multiple independent variables.

9. What is the main difference between Simple and Multiple Linear Regression?
  - Multiple Linear Regression contains multiple independent variables.
  - Simple Linear Regression contains only one independent variable.

10. What are the key assumptions of Multiple Linear Regression?
  - Relationship between all x and y should be linear.
  - Constant variance of errors.
  - No significant outliers or influential points.
  - Predictors are not highly correlated.

11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
  - Heteroscedasticity means the variance of the error terms is not constant across all levels of the independent variables.
  - Effects on Multiple Linear Regression:
    - Coefficient estimates remain unbiased
    - Standard errors become unreliable
    - t-tests and p-values may be incorrect.

12.  How can you improve a Multiple Linear Regression model with high multicollinearity?
  - Remove highly correlated predictors
  - Combine variables
  - Center or standardize variables

13.  What are some common techniques for transforming categorical variables for use in regression models?
  - One hot encoding and label encoding

14. What is the role of interaction terms in Multiple Linear Regression?
  - They model combined effects of predictors
  - Allow non-additive relationships
  - Improve model accuracy when variables influence each other

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
  - In Simple Linear Regression, intercept is the expected value of Y when X = 0.
  - In Multiple Linear Regression, intercept is the expected value of Y when all independent variables equal 0.

16. What is the significance of the slope in regression analysis, and how does it affect predictions?
  - The slope measures the change in the dependent variable (Y) for a one-unit increase in the independent variable, holding other variables constant.

17. How does the intercept in a regression model provide context for the relationship between variables?
  - The intercept provides a baseline reference point in a regression model.

18. What are the limitations of using R² as a sole measure of model performance?
  - Always increases when more variables are added
  - Does not measure prediction accuracy
  - Poor for comparing models with different predictors

19. How would you interpret a large standard error for a regression coefficient?
  - A large standard error for a regression coefficient means:
    - The coefficient estimate is imprecise
    - There is high uncertainty about its true value
    - The coefficient is less likely to be statistically significant

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?
  - Look for non-constant spread of residuals
  - Why it’s important to address:
    - Causes biased standard errors
    - Leads to incorrect t-tests and p-values
    - Makes confidence intervals unreliable

21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?
  - The model explains a large portion of variance(high R²), but many predictors are not truly useful(adjusted R²).

22.  Why is it important to scale variables in Multiple Linear Regression?
  - Makes coefficients comparable across predictors.
  - Prevents variables with large units to dominate the model.
  - Improve numerical stability of calculations.

23. What is polynomial regression?
  - Polynomial regression is a regression technique where the relationship between the independent variable and the dependent variable is modeled as nth degree polynomial.

24. How does polynomial regression differ from linear regression?
  - Linear regression is used for linear relationship between independent and dependent variable.
  - Polynomial regression is used for curved relationship between independent and dependent variable.

25. When is polynomial regression used?
  - Polynomial regression is used when relationship between independent and dependent variable is curved.

26. What is the general equation for polynomial regression?
  - y = axⁿ + ... + bx² + cx + d

27. Can polynomial regression be applied to multiple variables?
  - Yes

28. What are the limitations of polynomial regression?
  - If more degrees are used, model becomes overfit.

29.  What methods can be used to evaluate model fit when selecting the degree of a polynomial?
  - Adjusted R² - Penalizes unnecessary higher-degree terms.
  - RMSE/MAE - Measures prediction error.

30.  Why is visualization important in polynomial regression?
  - Reveals non-linear pattern in data.
  - Helps choose the appropriate polynomial degree.
  - Shows overfitting or underfitting clearly.

31.  How is polynomial regression implemented in Python?
  - Transform input features into polynomial terms.
  - Apply linear regression on those features.
  - Predict the test dataset.
