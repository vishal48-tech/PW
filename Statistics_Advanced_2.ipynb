1. What is hypothesis testing in statistics?

2. What is the null hypothesis, and how does it differ from the alternative hypothesis?
  - Null hypothesis is default assumption which always contains equal to. Alernate hypothesis is to find evidence for default assumption to be wrong.

3. Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

4. What are Type I and Type II errors? Give examples of each.
  - Type 1: When Ho is true and we reject it. E.g. You reject H₀ (conclude the drug works), but actually the drug does not work.
  - Type 2: When Ho is false and we accept it. E.g. If the test fails to detect the effect (does not reject H₀) even though the drug does work.

```python
# 7. Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.
import numpy as np
from scipy.stats import norm

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

alpha = 0.05
mu = 52 # Ho
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data)
sample_size = len(sample_data)

z_stats = (sample_mean - mu) / (sample_std / np.sqrt(sample_size))
p_value = 2 * (1 - norm.cdf(abs(z_stats)))   # two-tailed

if p_value < alpha:
  print("Result: Reject the null hypothesis (Significant difference).")
else:
  print("Result: Fail to reject the null hypothesis (No significant difference).")
```

```python
# 8. Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 50
sigma = 5
n = 1000

data = np.random.normal(mu, sigma, n)

sample_mean = np.mean(data)
sample_std = np.std(data)
alpha = 0.05

z_critical = norm.ppf(1 - alpha/2)

margin_of_error = z_critical * (sample_std / np.sqrt(n))
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", (ci_lower, ci_upper))

plt.figure(figsize=(8, 4))
plt.hist(data, bins=20)
plt.title("Simulated Normal Distribution Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# 9. Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram.
# Explain what the Z-scores represent in terms of standard deviations from the mean.
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 5, 1000)

sample_mean = np.mean(data)
sample_std = np.std(data)

z_scores = (data - sample_mean) / sample_std

print("First 10 Z-scores:", z_scores[:10])
print("Mean:", sample_mean)
print("Std Dev:", sample_std)

plt.figure(figsize=(8, 4))
plt.hist(z_scores, bins=20)
plt.title("Histogram of Z-Scores (Standardized Data)")
plt.xlabel("Z-Score")
plt.ylabel("Frequency")
plt.show()

# Answer
# +1 Z score means value is 1 std away from mean in positive direction
# -2 Z score means value is 2 std away from mean in negative direction
```
